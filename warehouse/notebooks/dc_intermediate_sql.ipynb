```python
import pandas as pd
```

<p>Clean films.csv</p>

```python
################################################################################
# Bronze Layer
################################################################################

# Step 1: Load data
df_films = pd.read_csv(
    "../../courses/dc_intermediate_sql/datasets/csv/films.csv", header=None
)

################################################################################
# Silver Layer
################################################################################

# Step 1: Create header columns
df_films.columns = [
    "id",
    "title",
    "release_year",
    "country",
    "duration",
    "language",
    "certification",
    "gross",
    "budget",
]

# Step 2: Change dtype from float64 to int64
df_films["release_year"] = df_films["release_year"].astype("Int64")

# Step 3: Change dtype from float64 to int64
df_films["duration"] = df_films["duration"].astype("Int64")

display(df_films)
```

<p>Clean reviews.csv</p>

```python
################################################################################
# Bronze Layer
################################################################################

# Step 1: Load data
df_reviews = pd.read_csv(
    "../../courses/dc_intermediate_sql/datasets/csv/reviews.csv", header=None
)

################################################################################
# Silver Layer
################################################################################

# Step 1: Reload raw data and generate id column so the cell is safe to re-run
df_reviews.insert(0, "id", range(1, len(df_reviews) + 1))

# Step 2: Create header columns
df_reviews.columns = [
    "id",
    "film_id",
    "num_user",
    "num_critic",
    "imdb_score",
    "num_votes",
    "facebook_likes",
]

# Step 3: Change `num_user` dtype from float64 to int64
df_reviews["num_user"] = df_reviews["num_user"].astype("Int64")

# Step 4: Change `num_critic` dtype from float64 to int64
df_reviews["num_critic"] = df_reviews["num_critic"].astype("Int64")

# Step 5: Round `imdb_score` to 2 decimal places
df_reviews["imdb_score"] = df_reviews["imdb_score"].round(decimals=2)

# Display process
display(df_reviews)
```
